```python
import pandas as pd

# 📂 Charger le fichier source
df = pd.read_csv("../data/raw/BornesPropres.csv")

# 📌 Colonnes à conserver pour les deux pages
colonnes_a_garder = [
    "id_station_local",
    "nbre_pdc",
    "date_mise_en_service",
    "date_maj",
    "coordonneesXY",
    "puissance_nominale",
    "adresse_station",
    "nom_operateur",
    "nom_station",
    "consolidated_latitude",     
    "consolidated_longitude" 
]

# 🧹 Garder uniquement ces colonnes (en vérifiant qu'elles existent bien)
colonnes_existantes = [col for col in colonnes_a_garder if col in df.columns]
df_light = df[colonnes_existantes]

# 💾 Sauvegarder le fichier allégé
df_light.to_csv("../data/raw/BornesPropresLight.csv", index=False)

print(f"✅ Fichier BornesPropresLight.csv créé avec {df_light.shape[0]} lignes et {df_light.shape[1]} colonnes.")
print(f"Colonnes conservées : {colonnes_existantes}")
```

```python
import pandas as pd

# 📂 Chemin vers le fichier original
input_file = "../data/clean/BornesPropres_with_depart_v2.csv"

# 📂 Chemin vers le fichier allégé
output_file = "../data/clean/BornesPropres_with_depart_light.csv"

# 1️⃣ Charger le fichier original
df = pd.read_csv(input_file)

# 2️⃣ Ne garder que la colonne nécessaire
df_light = df[["num_departement"]]

# 3️⃣ Sauvegarder en CSV
df_light.to_csv(output_file, index=False, encoding="utf-8")

print(f"✅ Fichier allégé sauvegardé : {output_file}")
print(f"📊 Nombre de lignes : {len(df_light)}")
print(f"📄 Colonnes conservées : {df_light.columns.tolist()}")
```
