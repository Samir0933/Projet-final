```python
import pandas as pd
import os

def create_light_bornes_file():
    """
    Crée une version allégée du fichier BornesPropres.csv
    en ne gardant que les colonnes utilisées dans l'application
    """
    
    # Chemin du fichier original
    input_file = "../data/raw/BornesPropres.csv"
    output_file = "../data/raw/BornesPropreFloLight.csv"
    
    # Vérifier que le fichier original existe
    if not os.path.exists(input_file):
        print(f"❌ Erreur: Le fichier {input_file} n'existe pas")
        return
    
    try:
        print(f"📊 Lecture du fichier {input_file}...")
        
        # Colonnes à conserver (basées sur votre code Streamlit)
        # Dans votre code, vous utilisez uniquement 'date_mise_en_service'
        # J'ajoute quelques colonnes qui pourraient être utiles pour l'analyse géographique
        colonnes_a_garder = [
            'date_mise_en_service',           # Utilisée dans votre code
            'coordonneesXY',                  # Pour la géolocalisation
            'consolidated_longitude',          # Coordonnées nettoyées
            'consolidated_latitude',           # Coordonnées nettoyées
            'Geopoint',                       # Point géographique
            'nom_station',                    # Nom de la station
            'adresse_station',                # Adresse
            'puissance_nominale',             # Puissance de la borne
            'nbre_pdc',                       # Nombre de points de charge
            'implantation_station'            # Type d'implantation
        ]
        
        # Lire le fichier original avec seulement les colonnes nécessaires
        print("🔄 Chargement des données (colonnes sélectionnées seulement)...")
        df_original = pd.read_csv(input_file, usecols=colonnes_a_garder, low_memory=False)
        
        # Informations sur le fichier original
        print(f"📈 Fichier original:")
        print(f"   - Lignes: {len(df_original):,}")
        print(f"   - Colonnes conservées: {len(df_original.columns)}")
        print(f"   - Taille mémoire: {df_original.memory_usage(deep=True).sum() / 1024 / 1024:.1f} MB")
        
        # Nettoyer les données de dates
        print("🧹 Nettoyage des données...")
        df_light = df_original.copy()
        
        # Convertir la date en datetime puis la remettre au format string pour économiser l'espace
        df_light['date_mise_en_service'] = pd.to_datetime(
            df_light['date_mise_en_service'], 
            errors='coerce'
        ).dt.strftime('%Y-%m-%d')
        
        # Supprimer les lignes où la date est manquante (optionnel)
        # df_light = df_light.dropna(subset=['date_mise_en_service'])
        
        # Créer le répertoire de sortie s'il n'existe pas
        os.makedirs(os.path.dirname(output_file), exist_ok=True)
        
        # Sauvegarder le fichier allégé
        print(f"💾 Sauvegarde vers {output_file}...")
        df_light.to_csv(output_file, index=False)
        
        # Informations sur le fichier allégé
        taille_originale = os.path.getsize(input_file) / 1024 / 1024  # MB
        taille_allegee = os.path.getsize(output_file) / 1024 / 1024   # MB
        reduction = (1 - taille_allegee / taille_originale) * 100
        
        print(f"\n✅ Fichier allégé créé avec succès!")
        print(f"📊 Statistiques:")
        print(f"   - Fichier original: {taille_originale:.1f} MB")
        print(f"   - Fichier allégé: {taille_allegee:.1f} MB")
        print(f"   - Réduction: {reduction:.1f}%")
        print(f"   - Lignes: {len(df_light):,}")
        print(f"   - Colonnes: {len(df_light.columns)}")
        
        print(f"\n📋 Colonnes conservées:")
        for col in df_light.columns:
            print(f"   - {col}")
            
        print(f"\n🔧 Pour utiliser le fichier allégé dans votre code Streamlit,")
        print(f"   remplacez dans la fonction load_data():")
        print(f'   bornes = pd.read_csv("data/raw/BornesPropreFloLight.csv", low_memory=False)')
        
    except Exception as e:
        print(f"❌ Erreur lors du traitement: {str(e)}")
        return

def preview_light_file():
    """
    Affiche un aperçu du fichier allégé
    """
    output_file = "data/raw/BornesPropreFloLight.csv"
    
    if not os.path.exists(output_file):
        print(f"❌ Le fichier {output_file} n'existe pas. Exécutez d'abord create_light_bornes_file()")
        return
    
    print(f"👀 Aperçu du fichier allégé:")
    df = pd.read_csv(output_file)
    print(df.head())
    print(f"\nℹ️  Informations sur les données:")
    print(df.info())

if __name__ == "__main__":
    # Créer le fichier allégé
    create_light_bornes_file()
    
    # Afficher un aperçu (optionnel)
    print("\n" + "="*50)
    preview_light_file()
```
