```python
import pandas as pd
import plotly.express as px

file_path_borne = '../data/raw/BornesPropres.csv'
df_borne = pd.read_csv(file_path_borne)

print(df_borne.columns)

kpis = {
    "Nombre total de points de charge": df_borne['nbre_pdc'].sum(),
    "Puissance moyenne (kW)": round(df_borne['puissance_nominale'].mean(), 2),
    "Pourcentage de bornes gratuites": round((df_borne['gratuit'] == True).mean() * 100, 2)
}

for nom, valeur in kpis.items():
    print(f"{nom} : {valeur}")
```

```python
import pandas as pd

# Chargement des données
df = pd.read_csv('../data/raw/TMJA2019_with_geopoint_VF_Samir.csv')

# Trafic moyen (TMJA actualisé) global (sur l'année 2019)
trafic_moyen = df['TMJA_actualise'].mean()
print(f"Trafic moyen (TMJA actualisé) sur l'année 2019 : {trafic_moyen:.2f} véhicules/jour")

# Densité moyenne annuelle : TMJA_actualise divisé par longueur en km
df['longueur_km'] = df['longueur'] / 1000  # convertir en km
df['densite'] = df['TMJA_actualise'] / df['longueur_km']

densite_moyenne = df['densite'].mean()
print(f"Densité moyenne annuelle : {densite_moyenne:.2f} véhicules/km/jour")
```
